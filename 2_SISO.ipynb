```python
import numpy as np
import numpy.random as nr
from sklearn.linear_model import LinearRegression
```

```python


# Number of pilot symbols
N = 100
PSNR_dB = 20
PSNR = 10**(PSNR_dB/10)
```

```python
# Complex pilot symbols (normalized)
x_real = 2*nr.randint(2,size=N) - 1
x_imag = 2*nr.randint(2,size=N) - 1
x = np.sqrt(PSNR)*(x_real + 1j*x_imag)
x
```

```python
# True complex channel
h_true = 1.2 + 0.7j

# Complex AWGN
noise = (nr.randn(N) + 1j*nr.randn(N))/np.sqrt(2)
noise
```

```python
# Recieved signal
y = h_true*x + noise

# Convert to real-valued regression
X1 = np.column_stack((np.real(x), -np.imag(x)))
X2 = np.column_stack((np.imag(x),  np.real(x)))

X = np.vstack((X1, X2))
y_stack = np.hstack((np.real(y), np.imag(y)))
y_stack
```

Linear Regression

```python
model = LinearRegression(fit_intercept=False)
model.fit(X, y_stack)

# Estimated channel
h_est = model.coef_[0] + 1j * model.coef_[1]

print("True channel h   :", h_true)
print("Estimated channel:", h_est)
print("Estimation error :", np.abs(h_true - h_est)**2)
```
